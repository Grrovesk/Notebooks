Fix Path Errors

```python
import site
from pathlib import Path

try:
    target = next(Path(site.getsitepackages()[0]).rglob("spaces/zero/utils.py"))

    txt = Path(target).read_text()
    txt = txt.replace(
        "Path('/proc/self/map_files').iterdir()",
        "[]  # disabled for modal"
    )
    Path(target).write_text(txt)

    print("✅ spaces.zero patched inside Modal")
except Exception as e:
    print("⚠️ Patch skipped:", e)
```

Add share=True

```python
from pathlib import Path

app_file = Path("app.py")

content = app_file.read_text(encoding="utf-8")

content = content.replace(
    "demo.queue().launch()",
    "demo.queue().launch(share=True)"
)

app_file.write_text(content, encoding="utf-8")

print("Updated app.py: share=True added")
```
