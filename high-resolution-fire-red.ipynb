Share=True Patch

```python
# Modify app.py to add share=True in demo.launch()

from pathlib import Path

app_file = Path("app.py")  # change path if needed

# Read file
text = app_file.read_text(encoding="utf-8")

# Replace allowed_paths line with allowed_paths + share=True
old = 'allowed_paths=["examples"],'
new = 'allowed_paths=["examples"],\n        share=True,'

if old in text:
    text = text.replace(old, new)
    app_file.write_text(text, encoding="utf-8")
    print("✅ share=True added successfully!")
else:
    print("⚠️ Target line not found. Check app.py formatting.")

# Optional: print last lines to verify
print("\n--- Last lines of app.py ---\n")
print("\n".join(app_file.read_text(encoding="utf-8").splitlines()[-12:]))
```
